## 1. Теоретическое введение: ключевые понятия сверточных нейронных сетей

### 1.1. Задача классификации изображений
Классификация изображений – это задача отнесения входного изображения к одному из заранее определённых классов. В нашем случае датасет **Fruits‑360** содержит 131 класс фруктов и овощей (яблоко, банан, виноград и т.д.). Требуется построить модель, которая по фотографии плода предскажет его правильный вид.

### 1.2. Сверточные нейронные сети (CNN)
Сверточная нейронная сеть – это класс глубоких нейронных сетей, специально разработанный для обработки данных с сеточной структурой, таких как изображения. В отличие от полносвязных сетей (MLP), CNN сохраняет пространственную структуру входных данных и использует локальные связи, что позволяет эффективно распознавать визуальные паттерны.

### 1.3. Основные компоненты CNN
- **Сверточный слой (Convolutional Layer)** – извлекает локальные признаки с помощью обучаемых фильтров (ядер). Каждый фильтр скользит по изображению и вычисляет скалярное произведение, создавая карту признаков.
- **Функция активации** – вносит нелинейность (обычно ReLU), чтобы модель могла аппроксимировать сложные зависимости.
- **Пулинг (Pooling)** – уменьшает пространственные размеры карт признаков, обеспечивая инвариантность к сдвигам и снижение вычислительной нагрузки.
- **Полносвязные слои (Dense)** – выполняют классификацию на основе высокоуровневых признаков, извлечённых свёрточными слоями.

### 1.4. Регуляризация
Для предотвращения переобучения используются методы **Dropout** (случайное отключение нейронов) и **Batch Normalization** (нормализация по мини‑батчу). Dropout заставляет сеть не полагаться на отдельные нейроны, а BatchNorm ускоряет обучение и стабилизирует распределение входных данных каждого слоя.

### 1.5. Метрики качества
Для многоклассовой классификации используют:
- **Accuracy** – доля правильных ответов.
- **Precision, Recall, F1‑score** – рассчитываются для каждого класса и усредняются (macro или weighted).
- **Confusion matrix** – матрица ошибок, показывающая, какие классы чаще всего путаются.

### 1.6. Фиксация времени обучения
Для оценки производительности и воспроизводимости эксперимента важно фиксировать время, затраченное на обучение модели. В коде необходимо использовать модуль `time` для замера длительности обучения и вывода результата.

### Шаг 7. Построение модели CNN

Мы будем использовать архитектуру с тремя свёрточными блоками, BatchNormalization и Dropout для регуляризации.

```python
model = models.Sequential([
    # Первый свёрточный блок
    layers.Conv2D(32, (3,3), padding='same', kernel_initializer='he_normal', input_shape=(img_height, img_width, 3)),
    layers.BatchNormalization(),
    layers.Activation('relu'),
    layers.MaxPooling2D((2,2)),

    # Второй свёрточный блок
    layers.Conv2D(64, (3,3), padding='same', kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.Activation('relu'),
    layers.MaxPooling2D((2,2)),

    # Третий свёрточный блок
    layers.Conv2D(128, (3,3), padding='same', kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.Activation('relu'),
    layers.MaxPooling2D((2,2)),

    # Классификатор
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(len(class_names), activation='softmax')
])

# Вывод архитектуры
model.summary()
```

```python
    img_array = tf.expand_dims(img_array, 0)
    pred = model.predict(img_array)
    return class_names[np.argmax(pred[0])], np.max(pred[0])

# Пример
print(predict('apple.jpg'))
```

## 3. Теоретическая часть

### 3.1. Свёртка
Сверточный слой содержит набор обучаемых фильтров (ядер) размера $k \times k \times C_{in}$. Фильтр «скользит» по входу и в каждой позиции вычисляет взвешенную сумму пикселей окна плюс смещение:

$$y_{i,j} = \sum_{u=0}^{k-1}\sum_{v=0}^{k-1}\sum_{c=1}^{C_{in}} w_{u,v,c}\; x_{i+u,\,j+v,\,c} + b.$$

Результат одного фильтра — карта признаков; слой из $F$ фильтров выдаёт $F$ карт. Размер выхода по каждой оси равен $O = \left\lfloor \frac{N - k + 2P}{S} \right\rfloor + 1$, где $N$ — размер входа, $P$ — паддинг, $S$ — шаг. В работе используются ядра 3×3 и `padding='same'`: выход того же размера, что и вход. Число параметров слоя $(k \cdot k \cdot C_{in} + 1) \cdot F$ не зависит от размера изображения: например, у первого слоя (3×3, 32 фильтра, RGB-вход) их $(3 \cdot 3 \cdot 3 + 1) \cdot 32 = 896$.

Ключевые свойства свёртки: **локальная связность** (нейрон видит только небольшое окно — рецептивное поле), **разделение весов** (один фильтр для всех позиций) и **эквивариантность к сдвигу** (сдвиг объекта сдвигает и карту признаков). С глубиной рецептивное поле растёт, и признаки становятся всё более абстрактными.

### 3.2. Пулинг
Слой подвыборки уменьшает пространственный размер карт признаков. `MaxPooling2D` с окном 2×2 оставляет максимум каждого окна и вдвое уменьшает высоту и ширину. У пулинга нет обучаемых параметров; он сокращает объём вычислений и даёт небольшую инвариантность к сдвигам. В нашей модели размер карт меняется так: 100 → 50 → 25 → 12 → 6.

### 3.3. Функции активации
- **ReLU** $f(x) = \max(0, x)$ вносит нелинейность, дёшево вычисляется и не насыщается при $x > 0$, поэтому градиенты не затухают, как у сигмоиды.
- **Softmax** на выходе превращает вектор оценок $z$ в распределение вероятностей по $K$ классам: $p_i = \dfrac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$.

### 3.4. Функция потерь и оптимизатор
Для многоклассовой классификации с целочисленными метками используется **разреженная категориальная кросс-энтропия** `sparse_categorical_crossentropy`: $L = -\log p_y$, где $p_y$ — вероятность, которую модель дала истинному классу $y$ (усредняется по батчу). Веса обновляет оптимизатор **Adam**: для каждого параметра он хранит скользящие средние градиента и его квадрата и за счёт этого адаптирует шаг обучения (по умолчанию $\eta = 10^{-3}$).

### 3.5. Batch Normalization
Слой нормализует выходы предыдущего слоя по мини-батчу, а затем масштабирует и сдвигает их обучаемыми параметрами $\gamma$ и $\beta$:

$$\hat{x} = \frac{x - \mu_B}{\sqrt{\sigma_B^2 + \varepsilon}}, \qquad y = \gamma \hat{x} + \beta.$$

При обучении используются статистики текущего батча, при инференсе — скользящие средние, накопленные за обучение. BatchNorm стабилизирует распределение входов следующих слоёв, допускает больший шаг обучения и ускоряет сходимость. В модели он стоит между свёрткой и ReLU.

### 3.6. Dropout
Во время обучения Dropout случайно обнуляет долю $p$ нейронов (у нас $p = 0.5$ перед выходным слоем), а остальные умножает на $1/(1-p)$. Сеть не может полагаться на отдельные нейроны и учится более устойчивым признакам, что уменьшает переобучение. При инференсе Dropout отключён.

### 3.7. Ранняя остановка
`EarlyStopping` следит за метрикой на валидационной выборке (у нас `val_loss`) и прекращает обучение, если она не улучшается `patience` эпох подряд (у нас 3). С `restore_best_weights=True` модель возвращает веса лучшей эпохи.

### 3.8. Метрики качества
- **Accuracy** — доля правильных ответов.
- **Precision** класса — доля верных среди всех ответов «этот класс»: $\frac{TP}{TP + FP}$. **Recall** — доля найденных объектов класса: $\frac{TP}{TP + FN}$. **F1** — их гармоническое среднее: $\frac{2 \cdot P \cdot R}{P + R}$.
- **Macro avg** — среднее по классам без весов, **weighted avg** — среднее с весами, пропорциональными числу объектов класса.
- **Матрица ошибок** $M_{ij}$ — число объектов класса $i$, отнесённых к классу $j$: на диагонали верные ответы, вне её — ошибки.

### 5.2. Загрузка датасета

Архив скачивается функцией `tf.keras.utils.get_file`: она кэширует файл в `~/.keras/datasets` и при повторном запуске не качает его заново. Распаковываются только папки `Training` и `Test`: в архиве лежат ещё статьи и исходный код авторов, которые для работы не нужны.

````python
# Прямая ссылка на архив датасета (задание 1)
DATASET_URL = "https://github.com/Horea94/Fruit-Images-Dataset/archive/refs/heads/master.zip"

archive_path = pathlib.Path(tf.keras.utils.get_file("Fruit-Images-Dataset-master.zip", origin=DATASET_URL))
print(f"Архив: {archive_path} ({archive_path.stat().st_size / 2**20:.0f} МБ)")

# Распаковываем только Training и Test (один раз)
extract_root = archive_path.parent / "fruits-360"
data_dir = extract_root / "Fruit-Images-Dataset-master"
if not (data_dir / "Training").exists():
    prefixes = ("Fruit-Images-Dataset-master/Training/", "Fruit-Images-Dataset-master/Test/")
    with zipfile.ZipFile(archive_path) as zf:
        zf.extractall(extract_root, members=[m for m in zf.namelist() if m.startswith(prefixes)])

train_dir = data_dir / "Training"
test_dir = data_dir / "Test"
print(f"Папка обучения: {train_dir.exists()}")
print(f"Папка тестирования: {test_dir.exists()}")
````

### 5.3. Создание наборов данных

`image_dataset_from_directory` строит `tf.data.Dataset` по структуре папок: имя подпапки — метка класса, классы нумеруются по алфавиту. Папка `Training` делится на обучающую (80 %) и валидационную (20 %) части с одинаковым `seed`, поэтому части не пересекаются. Тестовая выборка — папка `Test`. Её не перемешиваем, чтобы порядок меток не менялся при построении отчёта и матрицы ошибок.

````python
BATCH_SIZE = 32
IMG_HEIGHT = 100
IMG_WIDTH = 100

# Обучающая выборка: 80 % папки Training
train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
)

# Валидационная выборка: оставшиеся 20 % папки Training
val_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
)

# Тестовая выборка: папка Test, без перемешивания
test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    shuffle=False,
)

class_names = train_ds.class_names
assert class_names == test_ds.class_names, "Классы в Training и Test не совпадают"
NUM_CLASSES = len(class_names)
print(f"\nКоличество классов: {NUM_CLASSES}")
print(f"Примеры классов: {class_names[:5]}")
````

### 5.6. Нормализация и оптимизация конвейера

Изображения загружаются как `float32` со значениями 0..255. Нормализацию к диапазону 0..1 выполняет слой `Rescaling(1/255)`, встроенный в модель (раздел 6), то есть используется встроенная нормализация. Сохранённая модель сама нормализует вход, поэтому при инференсе этот шаг невозможно забыть. Ниже показано действие слоя на одном батче.

Для ускорения обучения наборы кэшируются в памяти (`cache`), обучающий набор перемешивается (`shuffle`), а следующий батч готовится, пока обрабатывается текущий (`prefetch`). Кэш хранится в `uint8`: пиксели — целые числа 0..255, так что потерь нет, а памяти нужно в 4 раза меньше — около 2,7 ГБ вместо 11 ГБ, которые не поместились бы в оперативную память Colab.

````python
# Диапазон значений до и после нормализации слоем Rescaling
images, _ = next(iter(train_ds))
normalized = layers.Rescaling(1.0 / 255)(images)
print(f"До нормализации:    dtype={images.dtype.name}, min={images.numpy().min():.0f}, max={images.numpy().max():.0f}")
print(f"После Rescaling:    dtype={normalized.dtype.name}, min={normalized.numpy().min():.2f}, max={normalized.numpy().max():.2f}")

AUTOTUNE = tf.data.AUTOTUNE


def to_uint8(images, labels):
    # Пиксели — целые 0..255, поэтому переход к uint8 без потерь, а кэш в 4 раза меньше
    return tf.cast(images, tf.uint8), labels


# Кэширование, перемешивание и предварительная выборка
train_ds = train_ds.map(to_uint8, num_parallel_calls=AUTOTUNE).cache().shuffle(1000, seed=SEED).prefetch(AUTOTUNE)
val_ds = val_ds.map(to_uint8, num_parallel_calls=AUTOTUNE).cache().prefetch(AUTOTUNE)
test_ds = test_ds.map(to_uint8, num_parallel_calls=AUTOTUNE).cache().prefetch(AUTOTUNE)
````

## 6. Построение модели (задание 3)

За основу взята архитектура из методических указаний: свёрточные блоки «Conv2D 3×3 → BatchNormalization → ReLU → MaxPooling 2×2» и классификатор «Flatten → Dense(512, ReLU) → Dropout(0.5) → Dense(131, softmax)». Изменения:

| | Базовая модель | Модель этой работы |
|---|---|---|
| Нормализация входа | нет | слой `Rescaling(1/255)` внутри модели |
| Свёрточные блоки (фильтры) | 3 блока: 32 → 64 → 128 | **4 блока: 32 → 64 → 128 → 256** |
| Выход перед Flatten | 12×12×128 = 18 432 | 6×6×256 = 9 216 |
| Параметров | ≈ 9,60 млн | ≈ 5,18 млн |

Зачем это сделано:
- **Четвёртый свёрточный блок** увеличивает глубину и рецептивное поле: признаки последнего блока охватывают почти весь плод, что помогает различать похожие сорта. Дополнительный пулинг вдвое уменьшает выход Flatten, поэтому Dense(512), самый тяжёлый слой сети, тоже становится вдвое меньше. Модель глубже базовой, но почти в два раза легче.
- **`Rescaling` внутри модели** — встроенная нормализация (см. п. 5.6).

````python
def conv_block(filters):
    """Свёрточный блок базовой архитектуры: Conv2D → BatchNorm → ReLU → MaxPooling."""
    return [
        layers.Conv2D(filters, (3, 3), padding="same", kernel_initializer="he_normal"),
        layers.BatchNormalization(),
        layers.Activation("relu"),
        layers.MaxPooling2D((2, 2)),
    ]


model = models.Sequential([
    layers.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 3)),
    layers.Rescaling(1.0 / 255),     # встроенная нормализация: 0..255 → 0..1

    *conv_block(32),                 # 100×100 → 50×50
    *conv_block(64),                 # 50×50 → 25×25
    *conv_block(128),                # 25×25 → 12×12
    *conv_block(256),                # 12×12 → 6×6 (добавленный четвёртый блок)

    # Классификатор
    layers.Flatten(),
    layers.Dense(512, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation="softmax"),
], name="fruits360_cnn")

# Вывод архитектуры
model.summary()
````

````python
# Схема модели (в Colab библиотеки pydot и graphviz уже установлены)
tf.keras.utils.plot_model(model, show_shapes=True, show_layer_activations=True, dpi=60)
````

````python
model.compile(
    optimizer="adam",                        # адаптивный оптимизатор, lr = 0.001
    loss="sparse_categorical_crossentropy",  # метки — целые номера классов
    metrics=["accuracy"],
)
````

## 7. Обучение модели (задание 4)

Параметры обучения:
- до **20 эпох**, размер батча 32, оптимизатор Adam (lr = 0.001);
- **EarlyStopping**: `monitor='val_loss'`, `patience=3`, `restore_best_weights=True`. Обучение останавливается, если потери на валидации не улучшаются 3 эпохи подряд, и модель возвращает веса лучшей эпохи;
- **время обучения** замеряется функцией `time.time()` до и после `model.fit()` и выводится в секундах и минутах (обязательное требование).

````python
EPOCHS = 20

# Ранняя остановка, чтобы избежать переобучения
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True,
)

# Замер времени обучения (обязательное требование)
print("Начало обучения...")
start_time = time.time()

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=[early_stop],
    verbose=1,
)

end_time = time.time()
training_time = end_time - start_time

epochs_run = len(history.history["loss"])
best_epoch = int(np.argmin(history.history["val_loss"])) + 1
print(f"\nОбучение завершено за {training_time:.2f} секунд ({training_time / 60:.2f} минут)")
print(f"Эпох выполнено: {epochs_run} из {EPOCHS}, в среднем {training_time / epochs_run:.1f} с на эпоху")
print(f"Лучшая эпоха по val_loss: {best_epoch}")
````

````python
# Графики потерь и точности на обучающей и валидационной выборках
epochs_range = np.arange(1, epochs_run + 1)
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
panels = [("loss", "Потери (loss), логарифмическая шкала"), ("accuracy", "Точность (accuracy)")]

for ax, (metric, title) in zip(axes, panels):
    ax.plot(epochs_range, history.history[metric], color=C_BLUE, lw=2, marker="o", ms=6, label="обучение")
    ax.plot(epochs_range, history.history[f"val_{metric}"], color=C_ORANGE, lw=2, marker="o", ms=6, label="валидация")
    ax.axvline(best_epoch, color="#898781", lw=1, label=f"лучшая эпоха ({best_epoch})")
    ax.set(title=title, xlabel="Эпоха", xticks=epochs_range)
    ax.grid(axis="y")
    ax.legend()

# Логарифмическая шкала потерь с обычными подписями: 0.01, 0.02, 0.05, 0.1, ...
plain = FuncFormatter(lambda v, _: f"{v:g}")
axes[0].set_yscale("log")
axes[0].yaxis.set_major_formatter(plain)
axes[0].yaxis.set_minor_locator(LogLocator(subs=(2, 5)))
axes[0].yaxis.set_minor_formatter(plain)
plt.tight_layout()
plt.show()
````

## 9. Функция предсказания (задание 6)

`predict_image(img_path)` принимает путь к изображению и возвращает пару «класс, уверенность»:
1. `load_img` загружает файл и приводит его к размеру 100×100, как у обучающих данных;
2. `img_to_array` превращает изображение в массив 100×100×3 со значениями 0..255;
3. `expand_dims` добавляет размерность батча: модель ждёт вход формы (N, 100, 100, 3);
4. модель возвращает 131 вероятность; класс — `argmax`, уверенность — максимальная вероятность.

Отдельная нормализация не нужна: её выполняет слой `Rescaling` внутри модели. Функция проверяется на 5 случайных изображениях разных классов из тестовой выборки.

````python
def predict_image(img_path):
    """
    Загружает изображение, подготавливает его и возвращает предсказанный класс и уверенность.
    """
    img = tf.keras.utils.load_img(img_path, target_size=(IMG_HEIGHT, IMG_WIDTH))
    img_array = tf.keras.utils.img_to_array(img)   # массив 100×100×3, значения 0..255
    img_array = tf.expand_dims(img_array, 0)       # добавляем размерность батча
    predictions = model.predict(img_array, verbose=0)
    predicted_class = class_names[np.argmax(predictions[0])]
    confidence = float(np.max(predictions[0]))
    return predicted_class, confidence
````

````python
# 5 случайных изображений разных классов из тестовой выборки (seed фиксирован — набор воспроизводим)
rng = np.random.default_rng(SEED)
sample_paths = []
for cls in rng.choice(class_names, size=5, replace=False):
    files = sorted((test_dir / cls).glob("*.jpg"))
    sample_paths.append(files[rng.integers(len(files))])

rows = []
fig, axes = plt.subplots(1, 5, figsize=(16, 4))
for ax, path in zip(axes, sample_paths):
    predicted, conf = predict_image(path)
    true_class = path.parent.name
    rows.append({"Файл": f"{true_class}/{path.name}", "Истинный класс": true_class,
                 "Предсказанный класс": predicted, "Уверенность": conf})
    ax.imshow(tf.keras.utils.load_img(path))
    ax.set_title(f"Истинный: {true_class}\nПрогноз: {predicted} ({conf:.1%})", fontsize=9)
    ax.axis("off")
plt.tight_layout()
plt.show()

task6_results = pd.DataFrame(rows)
task6_results
````

```python
# ===== 9. Функция predict_image =====
def predict_image(img_path):
    """
    Загружает изображение, подготавливает его и возвращает предсказанный класс и уверенность.
    """
    img = tf.keras.utils.load_img(img_path, target_size=(IMG_HEIGHT, IMG_WIDTH))
    img_array = tf.keras.utils.img_to_array(img)   # массив 100×100×3, значения 0..255
    img_array = tf.expand_dims(img_array, 0)       # добавляем размерность батча
    predictions = model.predict(img_array, verbose=0)
    predicted_class = class_names[np.argmax(predictions[0])]
    confidence = float(np.max(predictions[0]))
    return predicted_class, confidence
```
